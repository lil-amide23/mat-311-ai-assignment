### Prompt: 
I have a .csv file about credit card fraud. Load this data using pandas and output the first five rows of the data. Before that, import any libraries needed to do EDA using python, like seaborn, matplotlib, or numpy (as well as pandas)

### AI: ChatGPT‑5.6 Luna (I believe)

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization theme
sns.set_theme(style="whitegrid")

# Load dataset
df = pd.read_csv("card_transdata.csv")

# Display first five rows
df.head()
```

### Prompt:
Add shape, info, and description outputs of the dataset using pandas methods

### AI: ChatGPT‑5.6 Luna (I believe)

```python
# Display dataset dimensions
print("Dataset Shape:")
print(df.shape)

# Display dataset information
print("\nDataset Info:")
df.info()

# Display descriptive statistics
print("\nDataset Description:")
display(df.describe())
```

### Prompt: 
Write code to figure out how many missing values there are and drop them. Save the new dataframe as a new variable with "cleaned" in the name. Then describe the new data

### AI: ChatGPT‑5.6 Luna (I believe)

```python
# Count missing values in each column
print("Missing Values per Column:")
print(df.isnull().sum())

# Calculate total missing values
print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

# Drop rows containing missing values and save as new dataframe
cleaned_df = df.dropna()

# Describe the cleaned dataset
print("\nCleaned Dataset Shape:")
print(cleaned_df.shape)

print("\nCleaned Dataset Description:")
display(cleaned_df.describe())
```

### Prompt: 
Here is the data. In the following prompts, I will input instructions regarding data visualization of columns or rows. First, write code to make a countplot of the fraud column (which has value 1 or 0). Title the plot "Fraud Class Distribution"

### AI: ChatGPT‑5.6 Luna (I believe)

```python
# Countplot of fraud classes
plt.figure(figsize=(8, 5))
sns.countplot(data=cleaned_df, x="fraud")

# Add title and labels
plt.title("Fraud Class Distribution")
plt.xlabel("Fraud (0 = Legitimate, 1 = Fraudulent)")
plt.ylabel("Number of Transactions")

plt.show()
```

### Prompt:
Do a bivariate visualization using a countplot of if the customer used their pin number or not. It needs to be green and red on if the transaction was non-fraud or fraud, respectively.

### AI: ChatGPT‑5.6 Luna (I believe)

```python
plt.figure(figsize=(8, 6))

sns.countplot(
    data=cleaned_df,
    x="used_pin_number",
    hue="fraud",
    palette={0: "green", 1: "red"},
    edgecolor="black"
)

plt.title("PIN Usage by Fraud Status")
plt.xlabel("Used PIN Number")
plt.ylabel("Transaction Count")
plt.legend(title="Fraud Status", labels=["Non-Fraud", "Fraud"])

plt.tight_layout()
plt.show()
```

### Prompt:
Now do a similar plot but for percentages of fradulent transactions whether a PIN was used. Use a stacked bar plot

### AI: ChatGPT‑5.6 Luna (I believe)

```python
pin_fraud = pd.crosstab(
    cleaned_df["used_pin_number"],
    cleaned_df["fraud"],
    normalize="index"
) * 100

pin_fraud.plot(
    kind="bar",
    stacked=True,
    color=["green", "red"],
    edgecolor="black",
    figsize=(8, 6)
)

plt.title("Fraud Percentage by PIN Usage")
plt.xlabel("Used PIN Number")
plt.ylabel("Percentage of Transactions (%)")
plt.legend(title="Fraud Status", labels=["Non-Fraud", "Fraud"])

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Prompt:
Create a scatter plot to show if a transaction was fraud or not, with ratio_to_median_purchase_price on the y axis and distance_from_home on the x axis. Color the dots green or red for good or fraud

### AI: ChatGPT‑5.6 Luna (I believe)

```python
from matplotlib.lines import Line2D

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=cleaned_df,
    x="distance_from_home",
    y="ratio_to_median_purchase_price",
    hue="fraud",
    palette={0: "green", 1: "red"},
    alpha=0.6,
    legend=False
)

legend_elements = [
    Line2D([0], [0], marker='o', color='w',
           markerfacecolor='green', markersize=8, label='Non-Fraud'),
    Line2D([0], [0], marker='o', color='w',
           markerfacecolor='red', markersize=8, label='Fraud')
]

plt.title("Fraud vs. Non-Fraud Transactions")
plt.xlabel("Distance From Home")
plt.ylabel("Ratio to Median Purchase Price")
plt.legend(handles=legend_elements, title="Fraud Status")

plt.tight_layout()
plt.show()
```
